### 斐波那契（Fibonacci）函式定義

這裡提供幾種常見的斐波那契數列計算方法，以供後續使用。

1.  **`fibonacci_loop(n)`**: 使用迴圈迭代方式計算第 n 個斐波那契數。
2.  **`fibonacci_generator(limit)`**: 使用生成器產生斐波那契數列，直到達到指定限制。
3.  **`fibonacci_matrix(n)`**: 使用矩陣乘法（快速冪）計算第 n 個斐波那契數，效率較高。
4.  **`fibonacci_memo(n)`**: 使用遞迴和記憶化（memoization）計算第 n 個斐波那契數，避免重複計算。

```python
def fibonacci_loop(n):
    a, b = 0, 1
    for _ in range(n):
        a, b = b, a + b
    return a

def fibonacci_generator(limit):
    a, b = 0, 1
    while a < limit:
        yield a
        a, b = b, a + b

def fibonacci_matrix(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1

    # Matrix multiplication helper
    def multiply_matrices(M1, M2):
        a = M1[0][0] * M2[0][0] + M1[0][1] * M2[1][0]
        b = M1[0][0] * M2[0][1] + M1[0][1] * M2[1][1]
        c = M1[1][0] * M2[0][0] + M1[1][1] * M2[1][0]
        d = M1[1][0] * M2[0][1] + M1[1][1] * M2[1][1]
        return [[a, b], [c, d]]

    # Matrix power helper (exponentiation by squaring)
    def matrix_power(M, p):
        res = [[1, 0], [0, 1]] # Identity matrix
        M_copy = [[M[0][0], M[0][1]], [M[1][0], M[1][1]]]
        while p > 0:
            if p % 2 == 1:
                res = multiply_matrices(res, M_copy)
            M_copy = multiply_matrices(M_copy, M_copy)
            p //= 2
        return res

    # The Fibonacci matrix
    F = [[1, 1], [1, 0]]

    # Calculate F^(n-1)
    Fn_1 = matrix_power(F, n - 1)

    # The nth Fibonacci number is Fn_1[0][0]
    return Fn_1[0][0]

# Memoization dictionary
fib_cache = {}
def fibonacci_memo(n):
    if n in fib_cache:
        return fib_cache[n]
    if n <= 1:
        return n
    else:
        result = fibonacci_memo(n-1) + fibonacci_memo(n-2)
        fib_cache[n] = result
        return result

print("斐波那契函式已定義完成。")
```
